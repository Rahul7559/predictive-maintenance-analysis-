## Feature Engineeriing

### Creating Temperature Difference

```python

df["Temperature Difference [K]"] = (     
    df["Process temperature [K]"] - df["Air temperature [K]"]
)
```

```python
df["Temperature Difference [K]"] = df["Process temperature [K]"] - df["Air temperature [K]"]
```

```python
df[[
    "Air temperature [K]",
    "Process temperature [K]",
    "Temperature Difference [K]"
]].head()
```

### Create Power Proxy

```python
df["Power Proxy"] = (
    df["Rotational speed [rpm]"] * df["Torque [Nm]"]

)
```

```python
df[[
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Power Proxy"
]].head()             # This is a proxy/relative measure, not actual electrical power in watts.
```

### Create tool Wear Category

```python
df["Tool Wear Category"] = pd.cut(
    df["Tool wear [min]"],
    bins=[-1,100,200,float("inf")],
    labels=["Low","Medium","High"]
)
```

```python
df["Tool Wear Category"].value_counts()
```

```python
df.head() # checking the new dataset
```

```python
df.shape
```

```python
df.columns.tolist()
```

```python
target = "Machine failure"  # Defining the Target

y = df[target]

print(y.value_counts())
```

```python
features = [      # 
    
    "Type",
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]",
    "Temperature Difference [K]",
    "Power Proxy"
]

X = df[features]

X.head()    
```

```python
X = pd.get_dummies(X, columns=["Type"], drop_first=True) # Converting Type into numbers

X.head()
```

```python
X.columns
```

### Split the data

```python
# Using 80% for Training and 20% for Testing

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

### Scale the Features

```python
# Preparing the data for Logistic Regression Model 

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()                       
                                                #  we use fit_transform() only on training data and transform()
X_train_scaled = scaler.fit_transform(X_train)  #  on testing data ... This prevents informationn from the test set 
X_test_scaled = scaler.transform(X_test)        #  leaking into the training process..
```

### Train Logistic Regression

```python
# Training the Model

from sklearn.linear_model import LogisticRegression

model_lr = LogisticRegression(
    max_iter=1000,
    random_state=42
)

model_lr.fit(X_train_scaled, y_train)
```

### Make Predications

```python
y_pred_lr = model_lr.predict(X_test_scaled)
```

### Evaluate the model

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

accuracy = accuracy_score(y_test, y_pred_lr)
precision = precision_score(y_test, y_pred_lr)
recall = recall_score(y_test, y_pred_lr)
f1 = f1_score(y_test, y_pred_lr)

print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1 Score :", f1)
```

### Detailed classification report

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred_lr))
```

### Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm = confusion_matrix(y_test, y_pred_lr)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Failure", "Failure"]
)

disp.plot()
plt.title("Logistic Regression - Confusion Matrix")
plt.show()
```

### Train Decision Tree

```python
# Lets see whether a nonlinear model performs differently.

from sklearn.tree import DecisionTreeClassifier

model_dt = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

model_dt.fit(X_train, y_train)
```

### Make predictions

```python
y_pred_dt = model_dt.predict(X_test)
```

### Evaluate

```python
print(classification_report(y_test, y_pred_dt))
```

### Confusion Matrix

```python
cm_dt = confusion_matrix(y_test, y_pred_dt)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_dt,
    display_labels=["No Failure", "Failure"]
)

disp.plot()
plt.title("Decision Tree - Confusion Matrix")
plt.show()
```

### Compare the two models

```python
#  comparing Logistic Regression and Decision Tree

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

model_comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Decision Tree"],
    "Accuracy": [
        accuracy_score(y_test, y_pred_lr),
        accuracy_score(y_test, y_pred_dt)
    ],
    "Precision": [
        precision_score(y_test, y_pred_lr),
        precision_score(y_test, y_pred_dt)
    ],
    "Recall": [
        recall_score(y_test, y_pred_lr),
        recall_score(y_test, y_pred_dt)
    ],
    "F1 Score": [
        f1_score(y_test, y_pred_lr),
        f1_score(y_test, y_pred_dt)
    ]
})

model_comparison
```
